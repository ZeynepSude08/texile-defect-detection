```python
from rfdetr_plus import RFDETR2XLarge

DATASET_DIR = "data/coco_dataset"
OUTPUT_DIR = "runs/rfdetr_2xl_default"

EPOCHS = 50
BATCH_SIZE = 1
GRAD_ACCUM_STEPS = 16  

def main():
    model = RFDETR2XLarge(num_classes=1)

    model.train(
        dataset_dir=DATASET_DIR,
        output_dir=OUTPUT_DIR,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        grad_accum_steps=GRAD_ACCUM_STEPS,
        tensorboard=True,
        early_stopping=True,
    )

    print(f"\nTraining finished. Outputs saved to: {OUTPUT_DIR}")

if __name__ == "__main__":
    main()
```
